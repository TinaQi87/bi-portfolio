## 🔧 Part 2: The Retry Decorator Pattern

**Problem:** Network calls fail. Databases timeout. APIs rate-limit.

**Solution:** Retry with exponential backoff.

**What's a Decorator?** A function that wraps another function to add behavior.

````python
import time

def retry_on_failure(max_retries: int = 3, delay: int = 5):
    """
    Decorator that retries a function on failure.
    
    Uses exponential backoff: wait longer after each failure.
    Attempt 1 fails → wait 5s
    Attempt 2 fails → wait 10s
    Attempt 3 fails → wait 20s → give up
    """
    def decorator(func):  # Takes the function to wrap
        def wrapper(*args, **kwargs):  # Replacement function
            last_exception = None
            
            for attempt in range(max_retries):
                try:
                    return func(*args, **kwargs)  # Try the original function
                except Exception as e:
                    last_exception = e
                    wait_time = delay * (2 ** attempt)  # Exponential: 5, 10, 20...
                    print(f"Attempt {attempt + 1} failed: {e}. Retrying in {wait_time}s...")
                    time.sleep(wait_time)
            
            raise last_exception  # All retries failed
        return wrapper
    return decorator

# HOW TO USE:
@retry_on_failure(max_retries=3, delay=2)
def fetch_data_from_api():
    """This function will automatically retry on failure."""
    # Simulating a flaky API
    import random
    if random.random() < 0.7:  # 70% chance of failure
        raise ConnectionError("API timeout")
    return {"data": "success!"}

# Test it (may take a few attempts)
try:
    result = fetch_data_from_api()
    print(f"Success: {result}")
except Exception as e:
    print(f"All retries failed: {e}")
````

### 🎯 Decorator Pattern Breakdown:

```python
@retry_on_failure(max_retries=3)
def my_function():
    pass

# Is equivalent to:
my_function = retry_on_failure(max_retries=3)(my_function)
```

**When to use in Data Engineering:**
- Database connections
- API calls
- S3 uploads/downloads
- Any network operation

## 📊 Part 3: Data Extraction (Sources → Bronze)

### 3.1 Database Extraction with Generators

**Problem:** Tables can have millions of rows. Loading all at once = memory crash.

**Solution:** Use **generators** to process in batches.

````python
import pandas as pd
from typing import Generator
from datetime import datetime

class DatabaseExtractor:
    """
    Extract data from databases in memory-efficient batches.
    
    Why a class?
    - Manages connection state (open/close)
    - Groups related functions together
    - Reusable across different tables
    """
    
    def __init__(self, db_type: str = "mysql"):
        """Constructor - runs when you create an instance."""
        self.db_type = db_type
        self.conn = None  # Will hold database connection
    
    def extract_table(self, table: str, batch_size: int = 10000) -> Generator[pd.DataFrame, None, None]:
        """
        Extract table data in batches using a GENERATOR.
        
        Generator vs Regular Function:
        - Regular: return [all_data]  → loads everything into memory
        - Generator: yield batch      → processes one batch at a time
        
        The 'yield' keyword makes this a generator.
        """
        # Get total row count
        total_rows = 100000  # Simulated
        
        offset = 0
        while offset < total_rows:
            # In real code: query = f"SELECT * FROM {table} LIMIT {batch_size} OFFSET {offset}"
            # df = pd.read_sql(query, self.conn)
            
            # Simulated batch
            df = pd.DataFrame({
                'id': range(offset, min(offset + batch_size, total_rows)),
                'value': ['data'] * min(batch_size, total_rows - offset)
            })
            
            # Add metadata - ALWAYS track where data came from!
            df['_extracted_at'] = datetime.now()
            df['_source_table'] = table
            
            yield df  # YIELD returns data but keeps function state
            
            offset += batch_size
            print(f"Extracted {min(offset, total_rows)}/{total_rows} rows")

# USING THE GENERATOR:
extractor = DatabaseExtractor("mysql")

# This loop processes one batch at a time - memory efficient!
for batch_num, batch_df in enumerate(extractor.extract_table("orders", batch_size=25000)):
    print(f"Batch {batch_num + 1}: {len(batch_df)} rows")
    # Here you would: upload to S3, process, etc.
    if batch_num >= 2:  # Stop after 3 batches for demo
        break
````

### 3.2 API Extraction

**Common patterns:** Pagination, rate limiting, authentication.

````python
import requests

class APIExtractor:
    """Extract data from REST APIs."""
    
    @staticmethod  # No 'self' needed - doesn't use instance state
    def extract(url: str, headers: dict = None) -> pd.DataFrame:
        """
        Fetch data from API and convert to DataFrame.
        
        Type hints (url: str) help with:
        - Documentation
        - IDE autocomplete
        - Error catching
        """
        print(f"Fetching: {url}")
        
        response = requests.get(url, headers=headers, timeout=30)
        response.raise_for_status()  # Raises exception if status != 200
        
        data = response.json()
        
        # Handle different JSON structures
        if isinstance(data, list):
            df = pd.DataFrame(data)
        elif isinstance(data, dict):
            # Try common API patterns
            for key in ['data', 'results', 'items']:
                if key in data and isinstance(data[key], list):
                    df = pd.DataFrame(data[key])
                    break
            else:
                # Flatten nested dict
                df = pd.json_normalize(data)
        
        df['_extracted_at'] = datetime.now()
        df['_source_api'] = url
        return df

# Example: Fetch exchange rates
df = APIExtractor.extract("https://api.exchangerate-api.com/v4/latest/USD")
print(df.head())
````
