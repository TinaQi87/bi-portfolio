# Environment Validation Notebook
This notebook tests all components of your data engineering environment.

## 1. Test MySQL Connection

```python
import mysql.connector

try:
    conn = mysql.connector.connect(
        host="mysql",
        user="devuser",
        password="devpassword",
        database="devdb"
    )
    cursor = conn.cursor()
    cursor.execute("SELECT VERSION()")
    version = cursor.fetchone()
    print(f"✅ MySQL Connected Successfully!")
    print(f"MySQL Version: {version[0]}")
    cursor.close()
    conn.close()
except Exception as e:
    print(f"❌ MySQL Connection Failed: {e}")
```

## 2. Test PostgreSQL Connection

```python
import psycopg2

try:
    conn = psycopg2.connect(
        host="postgres",
        user="devuser",
        password="devpassword",
        database="devdb"
    )
    cursor = conn.cursor()
    cursor.execute("SELECT version()")
    version = cursor.fetchone()
    print(f"✅ PostgreSQL Connected Successfully!")
    print(f"PostgreSQL Version: {version[0][:50]}...")
    cursor.close()
    conn.close()
except Exception as e:
    print(f"❌ PostgreSQL Connection Failed: {e}")
```

## 3. Test AWS CLI

```python
import subprocess

try:
    result = subprocess.run(['aws', '--version'], capture_output=True, text=True)
    print(f"✅ AWS CLI Available!")
    print(result.stdout)
except Exception as e:
    print(f"❌ AWS CLI Failed: {e}")
```

## 4. Test AWS Credentials (List Regions)

```python
try:
    result = subprocess.run(
        ['aws', 'ec2', 'describe-regions', '--query', 'Regions[0:3].RegionName', '--output', 'text'],
        capture_output=True, text=True
    )
    if result.returncode == 0:
        print(f"✅ AWS Credentials Working!")
        print(f"Sample Regions: {result.stdout.strip()}")
    else:
        print(f"⚠️ AWS Credentials Not Configured or No Permission")
        print(result.stderr)
except Exception as e:
    print(f"❌ AWS Test Failed: {e}")
```

## 5. Test Terraform

```python
try:
    result = subprocess.run(['terraform', '--version'], capture_output=True, text=True)
    print(f"✅ Terraform Available!")
    print(result.stdout)
except Exception as e:
    print(f"❌ Terraform Failed: {e}")
```

## 6. Test Git

```python
try:
    result = subprocess.run(['git', '--version'], capture_output=True, text=True)
    print(f"✅ Git Available!")
    print(result.stdout)
except Exception as e:
    print(f"❌ Git Failed: {e}")
```

## 7. Test Python Packages

```python
import sys

packages = ['jupyter', 'mysql.connector', 'psycopg2']
print(f"Python Version: {sys.version}\n")

for package in packages:
    try:
        __import__(package)
        print(f"✅ {package} installed")
    except ImportError:
        print(f"❌ {package} not found")
```
